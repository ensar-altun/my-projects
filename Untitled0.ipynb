```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
import os
import zipfile
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, classification_report
import seaborn as sns
from sklearn.metrics import roc_curve, auc
from sklearn.metrics import precision_recall_curve
from sklearn.metrics import average_precision_score
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
train_dir = '/content/drive/MyDrive/Colab Notebooks/veri/train'
test_dir = '//content/drive/MyDrive/Colab Notebooks/veri/test'
```

```python
# Eğitim ve test verilerini yükleme
train_datagen = ImageDataGenerator(rescale=1.0/255, validation_split=0.2)  # Normalize ve doğrulama bölümü oluştur
test_datagen = ImageDataGenerator(rescale=1.0/255)  # Sadece normalize

# Eğitim ve doğrulama veri setlerini ayırma
train_data = train_datagen.flow_from_directory(
    train_dir,
    target_size=(150, 150),
    batch_size=32,
    class_mode="categorical",
    subset="training"
)

validation_data = train_datagen.flow_from_directory(
    train_dir,
    target_size=(150, 150),
    batch_size=32,
    class_mode="categorical",
    subset="validation"
)

# Test verisini yükleme
test_data = test_datagen.flow_from_directory(
    test_dir,
    target_size=(150, 150),
    batch_size=32,
    class_mode="categorical"
)
```

```python
# Modelin oluşturulması
model = tf.keras.Sequential([
    tf.keras.layers.Conv2D(32, (3, 3), activation='relu', input_shape=(150, 150, 3)),
    tf.keras.layers.MaxPooling2D(2, 2),
    tf.keras.layers.Conv2D(64, (3, 3), activation='relu'),
    tf.keras.layers.MaxPooling2D(2, 2),
    tf.keras.layers.Conv2D(128, (3, 3), activation='relu'),
    tf.keras.layers.MaxPooling2D(2, 2),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(512, activation='relu'),
    tf.keras.layers.Dense(train_data.num_classes, activation='softmax')  # Son katman sınıflar sayısına göre ayarlanmalı
])

# Modeli derleme
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])
```

```python
history = model.fit(
    train_data,
    epochs=10,
    validation_data=validation_data,
    verbose=2
)
```

```python
test_loss, test_acc = model.evaluate(test_data)
print(f'Test accuracy: {test_acc}')
```

```python
# Test verisiyle modelin tahminlerini yapma
test_preds = model.predict(test_data, verbose=1)
test_preds = np.argmax(test_preds, axis=1)  # Sınıf tahminlerini al

# Gerçek etiketleri alma
test_labels = test_data.classes

# Confusion Matrix hesaplama
cm = confusion_matrix(test_labels, test_preds)

# Confusion Matrix görselleştirme
plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=test_data.class_indices.keys(), yticklabels=test_data.class_indices.keys())
plt.xlabel('Predicted Labels')
plt.ylabel('True Labels')
plt.title('Confusion Matrix')
plt.show()

# Classification Report (Precision, Recall, F1-Score)
report = classification_report(test_labels, test_preds, target_names=test_data.class_indices.keys())
print("Classification Report:\n", report)
```

```python
# Test verisi için sınıfların doğrusal tahminlerini elde etme
y_pred = model.predict(test_data)
fpr, tpr, _ = roc_curve(test_labels, y_pred[:, 1], pos_label=1)  # Binary sınıflar için 1. sınıfı seçiyoruz
roc_auc = auc(fpr, tpr)

# ROC eğrisini çizme
plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, color='blue', lw=2, label=f'ROC curve (area = {roc_auc:.2f})')
plt.plot([0, 1], [0, 1], color='gray', linestyle='--')  # Random model çizgisi
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Receiver Operating Characteristic (ROC) Curve')
plt.legend(loc='lower right')
plt.show()
```

```python
from sklearn.preprocessing import label_binarize
# Test verisinin gerçek etiketleri (One-Hot Encoding değil, düz sınıf etiketleri)
true_labels = test_data.classes  # Örn: [0, 1, 2, ...]

# Tahmin edilen olasılıkları alıyoruz
y_pred = model.predict(test_data)

# Sınıf isimlerini al
class_names = list(test_data.class_indices.keys())

# Etiketleri binary formata dönüştür (One-vs-Rest için gerekli)
true_labels_binary = label_binarize(true_labels, classes=np.arange(len(class_names)))

# Her sınıf için Precision-Recall eğrisi çiz
plt.figure(figsize=(20, 20))

for i, class_name in enumerate(class_names):
    # Her sınıf için precision ve recall hesapla
    precision, recall, _ = precision_recall_curve(true_labels_binary[:, i], y_pred[:, i])

    # Precision-Recall eğrisi çiz
    plt.plot(recall, precision, lw=2, label=f'Class {i}')  # Sınıf isimleri yerine indeksler

plt.xlabel('Recall')
plt.ylabel('Precision')
plt.title('Precision-Recall Curve (Multi-Class)')
plt.legend(title='Classes', loc='lower left')
plt.grid()
plt.show()
```

```python
print(f"Confusion Matrix:\n{cm}")
print(f"Classification Report:\n{report}")
print(f"ROC AUC: {roc_auc:.2f}")
```

```python
loss, accuracy = model.evaluate(test_data)

# Accuracy ve Error Rate hesaplama
error_rate = 1 - accuracy

print(f"Accuracy (Doğruluk): {accuracy * 100:.2f}%")
print(f"Error Rate (Hata Oranı): {error_rate * 100:.2f}%")
```
